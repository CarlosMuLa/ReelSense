```python
import boto3
import sagemaker
from sagemaker.huggingface import HuggingFaceModel

sagemaker_session = sagemaker.Session()
role = sagemaker.get_execution_role()

# Apunta al modelo que ya subiste a S3
model = HuggingFaceModel(
    model_data='s3://reelsense-dataset-mlops-carlos/model/compressed-model/reelsense-movie-embeddings.tar.gz',
    role=role,
    transformers_version='4.26',
    pytorch_version='1.13',
    py_version='py39',
    env={
        'HF_TASK': 'feature-extraction'
    }
)
```

```python
print("correcto")
```

```python
predictor = model.deploy(
    initial_instance_count=1,
    instance_type='ml.t2.medium',
    endpoint_name='final-reelsense-endpoint'
)

print("Endpoint deployado:", predictor.endpoint_name)
```
